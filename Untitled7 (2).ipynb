```python
from flask import Flask, render_template_string, request

app = Flask(__name__)


# LOAN ELIGIBILITY RULE ENGINE


def check_loan_eligibility(
    age,
    income,
    credit_score,
    loan_amount,
    existing_emi,
    employment,
    repayment_history
):

    score = 0
    reasons = []


    # HARD REJECTION RULES

    if age < 18 or age > 65:
        return "Not Eligible", 0, [
            "Applicant age must be between 18 and 65."
        ]

    if income <= 0:
        return "Not Eligible", 0, [
            "Monthly income must be greater than zero."
        ]

    if credit_score < 600:
        return "Not Eligible", 0, [
            "Credit score is below the minimum requirement of 600."
        ]


    # CREDIT SCORE


    if credit_score >= 750:
        score += 30
        reasons.append("Excellent credit score (+30)")

    elif credit_score >= 700:
        score += 25
        reasons.append("Good credit score (+25)")

    elif credit_score >= 650:
        score += 15
        reasons.append("Average credit score (+15)")

    else:
        score += 5
        reasons.append("Low credit score (+5)")


    # MONTHLY INCOME


    if income >= 50000:
        score += 20
        reasons.append("Monthly income is ₹50,000 or more (+20)")

    elif income >= 30000:
        score += 15
        reasons.append("Monthly income is ₹30,000 or more (+15)")

    elif income >= 20000:
        score += 10
        reasons.append("Monthly income is ₹20,000 or more (+10)")

    else:
        score += 5
        reasons.append("Low monthly income (+5)")


    # EMPLOYMENT


    if employment == "Permanent":

        score += 15
        reasons.append("Permanent employment (+15)")

    elif employment == "Self Employed":

        score += 10
        reasons.append("Self-employed (+10)")

    else:

        score += 5
        reasons.append("Temporary employment (+5)")


    # EMI TO INCOME RATIO

    emi_ratio = (existing_emi / income) * 100

    if emi_ratio <= 30:

        score += 20
        reasons.append(
            f"Low EMI burden ({emi_ratio:.1f}%) (+20)"
        )

    elif emi_ratio <= 40:

        score += 15
        reasons.append(
            f"Moderate EMI burden ({emi_ratio:.1f}%) (+15)"
        )

    elif emi_ratio <= 50:

        score += 5
        reasons.append(
            f"High EMI burden ({emi_ratio:.1f}%) (+5)"
        )

    else:

        score -= 10
        reasons.append(
            f"Very high EMI burden ({emi_ratio:.1f}%) (-10)"
        )


    # REPAYMENT HISTORY


    if repayment_history == "Good":

        score += 15
        reasons.append("Good repayment history (+15)")

    elif repayment_history == "Average":

        score += 10
        reasons.append("Average repayment history (+10)")

    else:

        reasons.append("Poor repayment history (+0)")


    # LOAN AMOUNT


    max_loan = income * 60

    if loan_amount <= max_loan:

        score += 10

        reasons.append(
            "Requested loan amount is within the income-based limit (+10)"
        )

    else:

        score -= 10

        reasons.append(
            "Requested loan amount is high compared with income (-10)"
        )


    # FINAL DECISION


    if score >= 75:

        decision = "Eligible"

    elif score >= 50:

        decision = "Review Required"

    else:

        decision = "Not Eligible"

    return decision, score, reasons



# HTML PAGE


HTML = """
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>Loan Eligibility System</title>
<style>
  :root {--teal:#0f3d3e; --gold:#f2b84b; --ink:#13262b; --line:#d9e1e3}
  * {box-sizing:border-box}
  body {margin:0; color:var(--ink); font:16px/1.5 "Segoe UI",Arial,sans-serif; background:#f4f1ea}
  .wrap {display:grid; grid-template-columns:minmax(300px,2fr) 3fr; min-height:100vh}

  aside {background:var(--teal); color:#e8f1f0; padding:48px 40px}
  .logo {font-weight:700; letter-spacing:.5px; color:var(--gold); margin:0 0 56px}
  aside h1 {font:700 38px/1.15 Georgia,serif; margin:0 0 16px; color:#fff}
  aside p {color:#b7cfcd; margin:0 0 40px}
  aside h3 {font-size:15px; margin:0 0 14px; color:#fff}
  .w {display:grid; grid-template-columns:125px 1fr 26px; gap:10px; align-items:center;
      font-size:14px; margin-bottom:10px}
  .t {height:8px; background:rgba(255,255,255,.15); border-radius:4px}
  .t i {display:block; height:100%; background:var(--gold); border-radius:4px}
  .w em {font-style:normal; text-align:right; color:var(--gold)}
  .cut {margin:28px 0 0; font-size:14px}

  main {padding:48px 40px; max-width:720px}
  h2 {font:700 24px Georgia,serif; margin:0 0 20px}
  form {display:grid; grid-template-columns:1fr 1fr; gap:18px 20px}
  label {display:block; font-size:14px; font-weight:600; margin-bottom:6px}
  input, select {width:100%; height:48px; padding:0 14px; font-size:16px; color:var(--ink);
                 background:#fff; border:2px solid var(--line); border-radius:10px}
  input:focus, select:focus {outline:0; border-color:var(--teal)}
  button {grid-column:1/-1; height:52px; border:0; border-radius:10px; cursor:pointer;
          background:var(--gold); color:var(--teal); font-size:17px; font-weight:700}
  button:hover {background:#e8a92f}

  .res {background:#fff; border-radius:16px; padding:28px; margin-bottom:36px;
        display:grid; grid-template-columns:150px 1fr; gap:28px; align-items:center;
        border-top:6px solid var(--c)}
  .ring {width:150px; height:150px; border-radius:50%; display:grid; place-items:center;
         background:conic-gradient(var(--c) calc(var(--p) * 1%), #e6ebec 0)}
  .ring b {width:112px; height:112px; border-radius:50%; background:#fff; display:grid;
           place-items:center; font:700 38px Georgia,serif}
  .res h2 {color:var(--c); margin:0 0 6px}
  .res ul {grid-column:1/-1; margin:0; padding:0; list-style:none; display:flex; flex-wrap:wrap; gap:8px}
  .res li {font-size:14px; padding:5px 12px; border-radius:20px; background:#eef3f3}
  .res li.neg {background:#fbe7e4; color:#9b2c20}
  .res li.pos {background:#e3f3ec; color:#17674a}
  .res small {color:#5d6f73}

  @media (max-width:800px) {
    .wrap {grid-template-columns:1fr}
    aside, main {padding:32px 20px}
    .logo {margin-bottom:28px}
    form, .res {grid-template-columns:1fr}
    .w {grid-template-columns:110px 1fr 26px}
  }
</style>
</head>
<body>
<div class="wrap">

  <aside>
    <p class="logo">Loan Eligibility System</p>
    <h1>Know where you stand before you apply.</h1>
    <p>Answer a few quick questions and get a decision, a score and the reasons behind it.</p>

    <h3>Points available for each factor</h3>
    {% for n, p in [("Credit score", 30), ("Monthly income", 20), ("EMI burden", 20),
                    ("Employment", 15), ("Repayment history", 15), ("Loan size", 10)] %}
      <div class="w"><span>{{ n }}</span><div class="t"><i style="width:{{ p * 3 }}%"></i></div><em>{{ p }}</em></div>
    {% endfor %}
    <p class="cut">A score of 75 or more is Eligible. 50 to 74 needs review. Below 50 is not eligible.</p>
  </aside>

  <main>

    {% if result %}
      {% set c = "#1d8a5f" if result == "Eligible" else ("#d08a00" if result == "Review Required" else "#c0392b") %}
      <section class="res" id="result" style="--c:{{ c }}; --p:{{ [[score or 0, 0]|max, 100]|min }}">
        <div class="ring"><b>{{ score if score is not none else "-" }}</b></div>
        <div>
          <h2>{{ result }}</h2>
          <small>Your eligibility score. This is a college-project rule-based system and does not
                 represent a real bank's lending criteria.</small>
        </div>
        <ul>
          {% for r in reasons %}
            <li class="{{ 'neg' if '(-' in r else ('pos' if '(+' in r else '') }}">{{ r }}</li>
          {% endfor %}
        </ul>
      </section>
    {% endif %}

    <h2>Your details</h2>
    <form method="POST">

      {% for name, label, hint in [
          ("age", "Age", "18 to 65"),
          ("income", "Monthly income (₹)", "50000"),
          ("credit_score", "Credit score", "300 to 900"),
          ("loan_amount", "Loan amount (₹)", "1000000"),
          ("existing_emi", "Existing monthly EMI (₹)", "0 if none")] %}
        <div>
          <label>{{ label }}</label>
          <input type="number" name="{{ name }}" min="0" placeholder="{{ hint }}"
                 value="{{ request.form.get(name, '') }}" required>
        </div>
      {% endfor %}

      <div>
        <label>Employment type</label>
        <select name="employment" required>
          <option value="">Select</option>
          {% for o in ["Permanent", "Self Employed", "Temporary"] %}
            <option {{ "selected" if request.form.get("employment") == o }}>{{ o }}</option>
          {% endfor %}
        </select>
      </div>

      <div>
        <label>Repayment history</label>
        <select name="repayment_history" required>
          <option value="">Select</option>
          {% for o in ["Good", "Average", "Poor"] %}
            <option {{ "selected" if request.form.get("repayment_history") == o }}>{{ o }}</option>
          {% endfor %}
        </select>
      </div>

      <button type="submit">Check my eligibility</button>
    </form>

  </main>
</div>
{% if result %}
<script>document.getElementById("result").scrollIntoView({block: "center"});</script>
{% endif %}
</body>
</html>
"""



# FLASK ROUTE


@app.route("/", methods=["GET", "POST"])
def home():

    result = None

    score = None

    reasons = []


    if request.method == "POST":

        try:

            age = int(
                request.form["age"]
            )

            income = float(
                request.form["income"]
            )

            credit_score = int(
                request.form["credit_score"]
            )

            loan_amount = float(
                request.form["loan_amount"]
            )

            existing_emi = float(
                request.form["existing_emi"]
            )

            employment = request.form[
                "employment"
            ]

            repayment_history = request.form[
                "repayment_history"
            ]


            result, score, reasons = (
                check_loan_eligibility(
                    age,
                    income,
                    credit_score,
                    loan_amount,
                    existing_emi,
                    employment,
                    repayment_history
                )
            )


        except ValueError:

            result = "Invalid Input"

            reasons = [
                "Please enter valid numerical values."
            ]


    return render_template_string(

        HTML,

        result=result,

        score=score,

        reasons=reasons
    )



# START FLASK


import threading

def run_flask():

    app.run(
        host="0.0.0.0",
        port=5000,
        debug=False,
        use_reloader=False
    )


thread = threading.Thread(
    target=run_flask
)

thread.start()



# COLAB PUBLIC LINK


from google.colab.output import eval_js

url = eval_js(
    "google.colab.kernel.proxyPort(5000)"
)

print()
print("==========================================")
print("   LOAN ELIGIBILITY SYSTEM IS RUNNING")
print("==========================================")
print()
print("OPEN THIS LINK:")
print(url)
```
